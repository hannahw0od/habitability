```julia
using SpeedyWeather
```

```julia
spectral_grid = SpectralGrid()
model = PrimitiveWetModel(spectral_grid)
```

```julia
planet1 = EarthAtmosphere(spectral_grid, pressure_reference = 100000.0)
planet2 = EarthAtmosphere(spectral_grid, pressure_reference = 50000.0)
planet3 = EarthAtmosphere(spectral_grid, pressure_reference = 200000.0)

# Define and initialise the model
model1 = PrimitiveWetModel(spectral_grid; atmosphere=planet1)
model2 = PrimitiveWetModel(spectral_grid; atmosphere=planet2)
model3 = PrimitiveWetModel(spectral_grid; atmosphere=planet3)

simulation1 = initialize!(model1)
simulation2 = initialize!(model2)
simulation3 = initialize!(model3)
```

```julia
run!(simulation1, period=Day(30))
run!(simulation2, period=Day(30))
run!(simulation3, period=Day(30))
```

```julia
using GLMakie
flux = simulation1.diagnostic_variables.physics.surface_air_density
GLMakie.heatmap(flux, title="Surface Air Density")
```

```julia
using GLMakie
flux2 = simulation2.diagnostic_variables.physics.surface_air_density
GLMakie.heatmap(flux2, title="Surface Air Density")
```
