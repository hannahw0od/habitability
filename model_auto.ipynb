Write a script that automatically simulates multiple times with different initial conditions

```julia
using SpeedyWeather

# Define grid and Earth's rotation for future use
spectral_grid = SpectralGrid()
Ω_earth = 7.2921159e-5

# Create an array of rotation rates as a function of Earth's
rot_arr = collect(range(0.5, stop=1.5, step=0.1)) .* Ω_earth
```

```julia
# Create a loop which:
    # Creates a netCDF output file
    # Adds relevant outputs to netCDF
    # Creates a planet with a specific rotation rate
    # Initialises the model
    # Runs a simulation for a specified number of days 

for (idx, i) in enumerate(rot_arr)
    fname = "output_rot_$(idx).nc"                     

    # new NetCDF output for this run
    output = NetCDFOutput(spectral_grid; filename=fname)

    add!(output, SpeedyWeather.SoilMoistureOutput())
    add!(output, SpeedyWeather.SoilTemperatureOutput())
    add!(output, SpeedyWeather.SurfaceTemperatureOutput())

    planet = Earth(spectral_grid, rotation=i)
    model = PrimitiveWetModel(spectral_grid; output=output, planet=planet)
    sim = initialize!(model)
    run!(sim, period=Day(2), output=true)
end
```
