Adding soil moisture and temperature to the netCDF output file

```julia
# First create NetCDFOutput

using SpeedyWeather
spectral_grid = SpectralGrid()
output = NetCDFOutput(spectral_grid)

# Add to netCDF output: 
# SoilMoistureOutput, SoilTemperatureOutput, SurfaceTemperatureOutput
add!(output, SpeedyWeather.SoilMoistureOutput())
add!(output, SpeedyWeather.SoilTemperatureOutput())
add!(output, SpeedyWeather.SurfaceTemperatureOutput())
output
```

```julia
# Earth’s rotation rate
Ω_earth = 7.2921159e-5  # rad/s

planet2 = Earth(spectral_grid, rotation = 2*Ω_earth)

# # Define and initialise the model
model = PrimitiveWetModel(spectral_grid; output, planet=planet2)
simulation = initialize!(model)
```

```julia
# Run simulation
run!(simulation, period=Year(1), output=true)
```
